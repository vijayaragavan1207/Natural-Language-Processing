```python
# NLP Task 8: Text Classification using Naive Bayes


import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)


data = {
    "text": [
        "I love this movie, it was amazing",
        "The movie was excellent and wonderful",
        "I really enjoyed this film",
        "This is a fantastic movie",
        "The acting was brilliant",
        "I hated this movie",
        "The movie was terrible",
        "This film was boring and disappointing",
        "I did not enjoy the movie",
        "The acting was very poor",
        "Amazing experience, I loved it",
        "Wonderful movie with great acting",
        "Worst movie I have ever seen",
        "Very disappointing and boring film",
        "Excellent story and great performance"
    ],
    "label": [
        "positive",
        "positive",
        "positive",
        "positive",
        "positive",
        "negative",
        "negative",
        "negative",
        "negative",
        "negative",
        "positive",
        "positive",
        "negative",
        "negative",
        "positive"
    ]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)


X = df["text"]
y = df["label"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english"
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("\nTraining feature shape:", X_train_tfidf.shape)
print("Testing feature shape:", X_test_tfidf.shape)


model = MultinomialNB()

model.fit(X_train_tfidf, y_train)


y_pred = model.predict(X_test_tfidf)


accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

new_text = [
    "The movie was absolutely wonderful",
    "This was a terrible and boring movie",
    "I loved the excellent performance",
    "The film was disappointing"
]

new_text_tfidf = vectorizer.transform(new_text)

predictions = model.predict(new_text_tfidf)

print("\nNew Text Predictions:")

for text, prediction in zip(new_text, predictions):
    print(f"{text} --> {prediction}")
```
